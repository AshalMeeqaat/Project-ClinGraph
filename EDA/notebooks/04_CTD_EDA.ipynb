```python
def inspect_ctd(file):
    with open(file, "r", encoding="utf-8") as f:
        for i, line in enumerate(f):
            if not line.startswith("#") and line.strip() != "":
                print(file.name)
                print("First data line:", i)
                print(line)
                print("-"*80)
                break
```

```python
inspect_ctd(CHEMICALS)

inspect_ctd(DISEASES)

inspect_ctd(CHEMICAL_DISEASE)
```

```python
def dataset_summary(df, name):
    print("="*70)
    print(name)
    print("="*70)

    print(f"Rows              : {len(df):,}")
    print(f"Columns           : {len(df.columns)}")
    print(f"Duplicate Rows    : {df.duplicated().sum():,}")

    print("\nMissing Values")
    print(df.isna().sum())

    print("\nData Types")
    print(df.dtypes)

    print("\n")
```

```python
dataset_summary(chemicals,"Chemicals")

dataset_summary(diseases,"Diseases")

dataset_summary(chem_disease,"Chemical-Disease")

dataset_summary(chem_gene,"Chemical-Gene")
```

```python
import pandas as pd
import os

EXPORT = r"C:\Users\Ashal.Meeqaat\Desktop\Medical_KG_Project\datasets\Processed\CTD\csv"
os.makedirs(EXPORT, exist_ok=True)

##################################################
# Chemicals
##################################################

chemical = chemicals[["ChemicalID","ChemicalName"]].drop_duplicates()
chemical.columns = ["id","name"]

chemical.to_csv(
    os.path.join(EXPORT,"Chemical.csv"),
    index=False
)

##################################################
# Diseases
##################################################

disease = diseases[["DiseaseID","DiseaseName"]].drop_duplicates()
disease.columns = ["id","name"]

disease.to_csv(
    os.path.join(EXPORT,"Disease.csv"),
    index=False
)

##################################################
# Chemical -> Disease
##################################################

chem_dis = chem_disease[[
    "ChemicalID",
    "DiseaseID"
]].dropna().drop_duplicates()

chem_dis.columns = ["source","target"]

chem_dis.to_csv(
    os.path.join(EXPORT,"Chemical_ASSOCIATED_WITH_Disease.csv"),
    index=False
)

##################################################

print("Chemicals:",len(chemical))
print("Diseases:",len(disease))
print("Relationships:",len(chem_dis))
```
