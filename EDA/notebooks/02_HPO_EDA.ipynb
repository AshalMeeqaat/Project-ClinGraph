### Missing Values Analysis

Several columns (qualifier, onset, frequency, sex, modifier) contain a large number of missing values. These fields are optional clinical annotations rather than mandatory attributes, so the missing values are expected and do not indicate poor data quality.

Core graph fields (database_id, disease_name, hpo_id) are almost completely populated, making the dataset suitable for knowledge graph construction.

```python
hpo_clean = hpo.dropna(
    subset=[
        "disease_name",
        "hpo_id",
        "evidence"
    ]
)
```

```python
import pandas as pd
import os

EXPORT = r"C:\Users\Ashal.Meeqaat\Desktop\Medical_KG_Project\datasets\Processed\HPO\csv"
os.makedirs(EXPORT, exist_ok=True)

# Remove rows that are completely empty
hpo = hpo.dropna(subset=["database_id", "disease_name", "hpo_id"])

#########################################################
# Disease Nodes
#########################################################

disease = (
    hpo[["database_id", "disease_name"]]
    .drop_duplicates()
    .rename(columns={
        "database_id":"id",
        "disease_name":"name"
    })
)

disease.to_csv(
    os.path.join(EXPORT,"Disease.csv"),
    index=False
)

#########################################################
# Phenotype Nodes
#########################################################

phenotype = (
    hpo[["hpo_id"]]
    .drop_duplicates()
    .rename(columns={
        "hpo_id":"id"
    })
)

# Use the ID as the name because this annotation file
# doesn't contain phenotype names.
phenotype["name"] = phenotype["id"]

phenotype.to_csv(
    os.path.join(EXPORT,"Phenotype.csv"),
    index=False
)

#########################################################
# Relationships
#########################################################

relations = (
    hpo[["database_id","hpo_id"]]
    .drop_duplicates()
    .rename(columns={
        "database_id":"source",
        "hpo_id":"target"
    })
)

relations.to_csv(
    os.path.join(EXPORT,"HAS_PHENOTYPE.csv"),
    index=False
)

#########################################################

print("Disease :", len(disease))
print("Phenotype :", len(phenotype))
print("Relationships :", len(relations))
```
